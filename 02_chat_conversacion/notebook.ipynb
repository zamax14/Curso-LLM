# 02 - Chat conversacional

En un chat, el modelo no recuerda automáticamente lo que pasó antes. La aplicación debe reenviar el historial relevante en cada llamada.


## Paso 0: helper para enviar mensajes

Ollama usa una lista de mensajes. Cada mensaje tiene un `role` y un `content`.


```python
import ollama

MODEL = "llama3.2"

def responder(mensajes):
    respuesta = ollama.chat(model=MODEL, messages=mensajes)
    return respuesta["message"]
```

## Paso 1: dos llamadas sin historial compartido

En la segunda llamada preguntamos por un dato que se dijo en la primera, pero no se lo reenviamos al modelo.


```python
r1 = responder([{"role": "user", "content": "Me llamo Martín."}])
print("Primera respuesta:")
print(r1["content"].strip())

r2 = responder([{"role": "user", "content": "¿Cómo me llamo?"}])
print("\nSegunda respuesta:")
print(r2["content"].strip())
```

## Paso 2: construir un historial

Ahora guardamos los mensajes y los reenviamos. El mensaje `system` define el comportamiento general del asistente.


```python
historial = [
    {
        "role": "system",
        "content": "Eres un asistente amable que recuerda datos del usuario.",
    },
    {
        "role": "user",
        "content": "Me llamo Martín y estudio ingeniería en computación.",
    },
]

r1 = responder(historial)
historial.append(r1)
historial.append({"role": "user", "content": "¿Cómo me llamo y qué estudio?"})

r2 = responder(historial)
print(r2["content"].strip())
```

## Paso 3: ver el historial como datos

Esta celda muestra que la conversación es solo una estructura de Python que nosotros controlamos.


```python
from pprint import pprint

pprint(historial)
```

## Paso 4: chat interactivo opcional

Ejecuta la última línea si quieres probar un chat en el notebook. Para terminar, escribe `salir`.


```python
def chat_interactivo():
    historial = [
        {
            "role": "system",
            "content": "Eres un asistente útil y conciso. Responde en español.",
        }
    ]

    while True:
        usuario = input("Tú: ")
        if usuario.lower() == "salir":
            break
        historial.append({"role": "user", "content": usuario})
        respuesta = responder(historial)
        historial.append(respuesta)
        print(f"Modelo: {respuesta['content'].strip()}\n")

# chat_interactivo()
```
