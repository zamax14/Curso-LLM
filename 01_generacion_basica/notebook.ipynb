# 01 - Generación básica con un LLM local

En este notebook probamos la unidad mínima de trabajo con un LLM: enviar un `prompt` y recibir texto generado. La idea es mantenerlo simple antes de hablar de chat, embeddings o agentes.


## Paso 0: importar Ollama y definir el modelo

`MODEL` centraliza el nombre del modelo. Si en tu computadora usas otro, por ejemplo `qwen2.5:1.5b`, cámbialo aquí y vuelve a ejecutar las celdas siguientes.


```python
import ollama

MODEL = "llama3.2"

def generar(prompt, options=None):
    respuesta = ollama.generate(
        model=MODEL,
        prompt=prompt,
        options=options or {},
    )
    return respuesta["response"].strip()
```

## Paso 1: generación simple

El modelo recibe una instrucción corta y devuelve una continuación. No hay memoria ni estructura especial: solo entrada y salida de texto.


```python
prompt = "Explica qué es un modelo de lenguaje en una sola oración."
print(generar(prompt))
```

## Paso 2: controlar variación con `temperature`

Una temperatura baja tiende a respuestas más repetibles. Una temperatura alta permite más variedad, aunque también puede producir respuestas menos consistentes.


```python
for temperatura in [0.0, 1.5]:
    print(f"Temperature = {temperatura}")
    for intento in range(1, 4):
        texto = generar(
            "Nombra un lenguaje de programación.",
            options={"temperature": temperatura},
        )
        print(f"{intento}. {texto}")
    print()
```

## Paso 3: mejorar el prompt

El prompt puede definir el rol, el público y la longitud esperada. Eso suele mejorar mucho la utilidad de la respuesta sin cambiar el modelo.


```python
prompt_ingenieria = """Eres un asistente para estudiantes de ingeniería en computación.
Explica qué es la recursividad usando analogías de la vida real.
Sé breve, usa no más de 3 oraciones."""

print(generar(prompt_ingenieria))
```

## Para experimentar

Cambia el tema, la temperatura o el límite de oraciones. La parte importante del ejercicio es observar cómo pequeñas variaciones en la entrada cambian la salida.
